# The Continue Assistant in VS Code

In this course we use [Continue](https://www.continue.dev), an open-source AI coding assistant, inside VS Code. Continue passes our requests to a large-language model (LLM) running on College infrastructure (the HX3 cluster), as configured in [00-setup.ipynb](00-setup.ipynb). We do not use commercial tools such as GitHub Copilot or Microsoft Copilot; if VS Code offers to install Copilot or sign in to GitHub, you can say no.

Using our own model has some practical consequences:
* There is no subscription: access is through your HX3 account and a token (API key), which expires after one hour but can be renewed indefinetly.
* Your code and prompts go to the College cluster, not to a third-party company.
* The models hosted on HX3 (for example, Qwen and Devstral) are generally smaller than the most advanced commercial models. As a result, it is important to check their outputs carefully and validate any generated code or analysis. Doing so will help you develop good practices for using AI tools effectively and critically.

Continue suggests code as you type (autocompletion), answers questions in a chat panel, and in *Agent* mode it can read files, create and edit files and run commands, asking your permission first. Because it integrates into your editor, it is much more powerful and easier to use for programming than a standalone chat window.

In this course, we will be using it in VS Code in relation to the Python programming language. However, the agent and models supports many programming languages.

In this notebook, we'll look at the Continue interface. In future notebooks, we'll look at how to use it to achieve specific tasks when programming.

## Implementation in VS Code

VS Code has many extensions that can be installed to add additional functionality. These can be accessed via the marketplace, which is accessible from the extensions tab on the left-hand side of the screen.

Continue is used through the **Continue extension** (publisher: Continue), installed from the marketplace. Its settings, including the connection to the model on HX3, are stored in Continue's `config.yaml` file, as described in [00-setup.ipynb](00-setup.ipynb).

## Autocompletion

When typing code in your editor, Continue can provide suggested completions for the current line or block of code. When a suggestion appears as grey text, you can accept it in its entirety by pressing **Tab**, accept only the next word by pressing **Ctrl + Right Arrow** (**Cmd + Right Arrow** on macOS), or reject it by pressing **Escape** or simply continuing to type.

For longer suggestions, you can accept an entire block of generated code by pressing **Tab**. Suggestions may take a few moments to appear because each suggestion requires a request to a model running on HX3.

## Working with highlighted code

Consider the following code cell as an example.

```python
def add(x, y):
    return x - y
```

When you highlight code in the editor, Continue shows a small hint next to the selection with two shortcuts:

* **Command (⌘) + L on Mac / Ctrl + L on Windows (Chat):** adds the highlighted code to the chat panel as context. You can then ask a question about it, for example *"This function is called `add`. Does what it returns match its name?"*. The code itself is not changed.

* **Command (⌘) + I on Mac / Ctrl + I on Windows (Edit):** opens a small input box at the selection where you type an instruction, for example *"fix this function so it adds the two numbers"*. Continue rewrites the highlighted code in place and shows the proposed change as a diff, with new lines on a green background and removed lines on a red background. You can accept or reject the change, so nothing is kept until you approve it.

Use **Chat** when you want an explanation or a second opinion, and **Edit** when you want Continue to change the code directly. In both cases, only the highlighted code is sent to the model, so selecting just the relevant lines gives more focused answers.

If you can't see the hint, check that the highlighted text is inside the code cell, that the Continue extension is active, and that no other extension uses the same shortcuts.

## Chat Interface

The Continue chat interface appears in the side panel. If it's not open, click the Continue icon in the activity bar on the left of the editor, or press ctrl/cmd + L. This will open a chat interface where you can ask questions about your code, or ask the agent to perform different tasks. 

### Context

You can give Continue context in two ways: highlight code in a file and press ctrl/cmd + L to send the selection, or type `@` in the chat box to attach other context, such as `@Files` followed by the start of a file name to attach a whole file. It is possible to add several items in this way. By providing appropriate context, you can get more relevant responses more quickly and with a lower token usage. Briefly explaining the relevance of each piece of context can also help explain the situation to the model and improve its responses.

**Task**: Try adding this whole notebook as context and asking Continue to give you a high-level overview of the file.

```python
def find_unique_characters(string):
    characters = set()
    for char in string:
        characters.add(char)

    characters = list(characters)
    characters.sort()

    return characters

print(find_unique_characters("hello"))
print(find_unique_characters("Imperial College London"))
```

### Modes

Continue has three modes, selected with the mode selector below the chat input box (or cycled with ctrl/cmd + .):
* **Agent**: Continue can use tools: read, create and edit files and run terminal commands. Before using a tool it asks for your permission, unless you have changed that tool's policy.
* **Chat**: Continue only answers questions; it does not use tools or modify files.
* **Plan**: Continue can read your code (read-only tools) and propose a plan, but it does not make changes. Once you agree with the plan, switch to Agent mode to carry it out.

How much Agent mode may do without asking is set per tool in Continue's tool settings, where each tool can be set to *Ask First*, *Automatic* or *Excluded*. Do not set tools that edit files or run commands to *Automatic* in this course. Agent and Plan modes need a model that supports tool use; otherwise Continue shows them as not supported.

It is also possible to give Continue extra capabilities (e.g. MCP servers) and to define reusable instructions ("rules" and "prompts"), but that is beyond the scope of this introduction.

### Models

Continue uses the model that we configured during setup, which is served by HX3. Models are listed in Continue's `config.yaml` file, each with one or more *roles* (`chat`, `edit`, `apply`, `autocomplete`). The current chat model is shown below the chat input box, and you can click on it to switch if more than one is configured.

Larger models generally give better answers, especially for complex reasoning and for agent tasks that require using tools, but respond more slowly. Very small models may struggle to use Continue's tools correctly, so if Continue describes an action instead of carrying it out, the model is a likely cause. Autocompletion works best with a model trained for code completion.

### Context Window

An agent's working memory is defined by its context window. This is its memory of the current conversation, the files it has looked at, any context and instructions. If the context window fills, the oldest parts of the conversation are dropped or compressed so that the rest fits, which can lead to the agent forgetting details.

When you prompt the agent, it will examine the contents of its context window and use that information to generate a response. As the window fills, this will increase the token usage of each prompt and the latency of the response.

The size of the context window is set by how the model is deployed on HX3 (and by `contextLength` in Continue's `config.yaml`), and is usually smaller than for commercial services, so it is worth keeping the context focused.

### Starting a New Session

It's often a good idea to start a new session when changing tasks due to the behaviour of the context window. This can minimise token usage and improve response times. However, this will also clear the context window, so you will need to provide any relevant context again. You can do this by clicking the `+` button at the top of the Continue panel.

### Revisiting Old Sessions

Continue keeps a history of your sessions, which you can open with the history button at the top of the Continue panel. You can click on a session to open it and continue the conversation. This means you can have several sessions in use and, if you want to revisit a task you have previously worked on, you can open the relevant session and continue the conversation with its history intact.

### Tokens and Usage

The use of LLMs is measured in tokens, which roughly corresponds to the number of words the model processes. (These tokens are completely unrelated to the access token or API key we need to generate to access the model). Commercial companies chage users for the number of tokens they use. We do not have any sort of limitation on the number of tokens used on HX3. But note that you are using a shared resource, you should be mindful of the use you make of it, as responses, including autocompletion suggestions, may be slower when many people are using it.

## Privacy and Security

It's important to be aware of the privacy and security implications of any Generative AI tool. Because the model runs on College infrastructure, your code and prompts are not sent to a commercial provider, which makes this set-up suitable for unpublished research code. They do, however, leave your computer and are processed on the HX3 cluster. Note that autocompletion sends the code around your cursor to the model as you type.

Some precautions still apply:
* Your Client_ID/Client_secret pair and your access token are equivalent to a password. Keep you 00-setup.ipynb and Continue's `config.yaml` private, and never paste them into a chat.
* In Agent mode, Continue can read and modify files and run commands on your computer. Keep tools that change files or run commands on *Ask First*, read each request before allowing it, and open in VS Code only the folders you want it to work on.
* Check that Continue only uses the HX3 models: in Continue's `config.yaml`, every model should have an `apiBase` on the HX3 server (`...hx3.hpc.ic.ac.uk/v1`), and the model dropdown below the chat box should list only HX3 models. Also turn off Continue's anonymous usage statistics: open VS Code Settings (**⌘ + comma** on Mac, **Ctrl + comma** on Windows, or the gear icon at the bottom left → *Settings*), search for "continue telemetry" and untick **Continue: Telemetry Enabled**.

Imperial provides [more information on protecting sensitive data in general](https://www.imperial.ac.uk/research-and-innovation/support-for-staff/scholarly-communication/research-data-management/data-storage-and-security/storing-sensitive-and-personal-data/).

## Reducing Plagiarism

Open-weight models such as Qwen were trained on large amounts of public code, including code under licenses that restrict duplication and/or modification. They can occasionally reproduce such code closely. Unlike some commercial services which have options hat claim blocking suggestions which match public code, our set-up has no such aspiration so the responsibility for checking lies with you. The fact that models can do this is an ethical issue with potential legal implications that have not been fully explored yet.

Even when not copied exactly, suggested code will still be inspired by the code in public repositories, including code under a license which prevents duplication and/or modification. This can still be considered an ethical issue, but most would consider it a lesser issue than outright duplication of the code.

The ethics and legality of generative AI of all sorts are a matter of debate in the context of plagiarism. As a result, there is not a "correct" stance to communicate to you, but it is an issue you should be aware of and consider.
